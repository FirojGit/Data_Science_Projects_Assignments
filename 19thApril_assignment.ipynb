```python
# Given a DataFrame with missing values: 
#•  Fill missing values in column A with the mean of the column. 
#•  Replace missing values in column B with the median. 
#•  Drop rows where column C has missing values. 
import pandas as pd
import numpy as np

data = {'A': [1, 2, np.nan, 4, 5], 
'B': [5, np.nan, np.nan, 8, 10], 
'C': ['a', 'b', np.nan, 'd', 'e']} 
df = pd.DataFrame(data) 
#print(df)
df["A"] = df["A"].fillna(df["A"].mean())
df["B"] = df["B"].fillna(df["B"].median())
df = df.dropna(subset= ["C"])
print(df)
```

```python
#Given a DataFrame of sales data: 
#•  Pivot the table to show total sales for each product across regions. 
#•  Add a new column to the pivot table that calculates the percentage of total sales for each 
#product.
    
data = {'Region': ['North', 'South', 'North', 'South', 'East'], 
'Product': ['A', 'A', 'B', 'B', 'C'], 
'Sales': [200, 150, 400, 300, 100]} 
df = pd.DataFrame(data) 
print(df)

df = df.pivot_table(values='Sales',index = 'Product' ,aggfunc='sum')
df['percent'] = (df['Sales']/df['Sales'].sum())*100

#df.pivot(index= 'Region', values = ['Sales', 'total sales'], columns ='Product' )

print(df)
```

```python
# Write a custom function that: 
#• Takes a numeric value as input and returns 'High' if the value is above the mean of 
#the column, otherwise 'Low'. 
#• Apply this function to a column of random integers and add the results as a new 
#column. 
import numpy as np
import pandas as pd
data = {'col1': np.random.randint(1,10,size=5)}
df = pd.DataFrame(data)
print(df)
def custom(num, mean):
    return "High" if num>mean else "Low"
    
mean = df['col1'].mean()
df["Result"] = df["col1"].apply(lambda x:custom(x,mean))
df
```

```python
#Create a Pandas Series of random integers between 1 and 100. Perform the following tasks: 
#1. Create a new series where all values that are greater than the 75th percentile of the 
#original series are replaced with the string 'High' and the rest are replaced with 'Low'. 
#2. Compute and display the cumulative sum of the series values

randVal= np.random.randint(1,101,size= 8)
S = pd.Series(randVal)
S

Sorted= np.sort(S)
n= len(Sorted)
pos= int(0.75*n)
percentile_75th = Sorted[pos]

new_series= S.apply(lambda x:"High" if x>percentile_75th else "Low")
print(new_series)

cum_sum = S.cumsum()
print(cum_sum)

```
