```python
# Write a program to find the second largest number in a list without using built-in 
#functions like max().

list1 = [12, 15, 1, 10, 10, 1]
largest = list1[0]
secnd_large = list1[0]

for i in list1:
    if i > largest:
        secnd_large = largest
        largest = i
    elif (i > secnd_large and i != largest):
        secnd_large = i

print("Second largest number is:", secnd_large)
```
