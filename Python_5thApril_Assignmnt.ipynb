```python
# Print numbers from 1 to 10. If a number is even, break the loop using a for loop
#and else clause

for num in range(1,11):
    if (num%2==0):
        break
    print(num)
    
    
```

```python
# Print numbers from 1 to 5, except 3 using a for loop and continue statement

for num in range(1,6):
    if(num==3):
        continue
    print(num)
```

```python
# Print numbers in a list until a negative number is encountered using 
#a while loop. 

list1 = [5,4,2,1,0,-1,-2,3]

i=0
while list1[i]>=0:
    print(list1[i])
    i += 1
```

```python
#Find the common elements in two lists using a for loop

list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

common = []
for i in list1:
    if i in list2:
        common.append(i)

print(common)    
```

```python
# Print the Fibonacci sequence up to the 10th term using a while loop

febo_list = [0,1]
for i in range(2,10):
    febo_list.append(febo_list[i-1]+ febo_list[i-2])
print(febo_list)
```

```python
# Print numbers divisible by 3 or 5 from 1 to 20 using a for loop

for i in range (1,21):
    if i%3==0 or i%5==0:
        print(i)
```

```python
# Find all prime numbers between 1 and 50 using nested for and if

for num in range(2,51):
    for i in range(2,num):
       if num%i ==0:
         break
    else:
        print(num)
```

```python
#  Find the first occurrence of a number in a list using a while loop

lis1 = ["Python", "code", '1' ,'2', '4',"hkkj"]

for i in lis1:
    if i.isdigit():
        print(i)
```

```python
#Calculate factorial of a number using a while loop

number = int(input("Enter number:"))

fact =1
for i in range (1,number):
    fact = fact*i
print(fact)
```

```python
#  Print patterns using nested for loops.

for i in range (1,6):
    for j in range(i):
            print('*', end='')
    print()
    
```

```python
for i in range (6,1,-1):
    for j in range(i):
            print('*', end='')
    print()
```
